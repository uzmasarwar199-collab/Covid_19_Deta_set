```python
df = pd.read_csv("archive (2).zip")
df.head()
```

```python
df.isnull().sum()
```

```python
df.fillna(0, inplace=True)
```

```python
# Fix data types properly
df['day'] = pd.to_datetime(df['day'])
df['time'] = pd.to_datetime(df['time'], errors='coerce')

# Ensure numeric columns don't have negative numbers
num_cols = ['Cases', 'Recovered', 'Deaths', 'Tests']
for col in num_cols:
    df[col] = df[col].clip(lower=0)
```
